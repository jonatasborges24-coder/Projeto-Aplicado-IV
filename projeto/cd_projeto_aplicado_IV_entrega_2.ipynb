# **Introdução**

A literatura portuguesa constitui um importante patrimônio cultural, sendo
fundamental para a formação linguística, crítica e interpretativa de estudantes e
pesquisadores. No entanto, o acesso estruturado a obras literárias e a análise contextual
de seus conteúdos ainda representam desafios em ambientes digitais.

Nesse contexto, o projeto Memoria Litteraria consiste em uma aplicação web
voltada à busca textual e análise de obras da literatura portuguesa. A plataforma permite
localizar trechos específicos, analisar o uso de palavras em diferentes contextos e
recomendar obras e autores com base no perfil de uso. Seu público-alvo inclui estudantes,
pesquisadores, professores e escritores.

Além das funcionalidades de busca e recomendação, o sistema coleta dados  de
interação dos usuários, como consultas realizadas, funcionalidades utilizadas e
avaliações. Esses dados, organizados ao longo do tempo, configuram um cenário
adequado para análise por meio de séries temporais.

Séries temporais consistem em observações ordenadas cronologicamente,
permitindo a identificação de padrões como tendência, sazonalidade e variações
aleatórias. A análise desses padrões possibilita compreender a dinâmica de uso da
plataforma, identificar comportamentos recorrentes e apoiar a tomada de decisão baseada
em dados.

Dessa forma, o problema abordado neste projeto consiste na necessidade de
compreender como o comportamento dos usuários se desenvolve ao longo do tempo,
permitindo aprimorar a experiência na plataforma e orientar melhorias fundamentadas em
evidências.  

A motivação do projeto está na integração entre tecnologia e literatura, com o
objetivo de tornar o acesso ao conhecimento literário mais eficiente, estruturado e
orientado por dados. Do ponto de vista acadêmico, a análise de séries temporais permite
compreender padrões dinâmicos de comportamento, ampliando o uso de técnicas de
Ciência de Dados no contexto educacional. Sob a perspectiva tecnológica, a análise
temporal das interações dos usuários possibilita identificar padrões de uso, períodos de
maior atividade e mudanças de interesse, contribuindo para a evolução contínua da
plataforma.

No âmbito social, o projeto alinha-se ao Objetivo de Desenvolvimento Sustentável
4 (ODS 4), ao promover o acesso à educação de qualidade por meio de uma ferramenta
digital aberta e acessível. O caráter extensionista do projeto se manifesta na
disponibilização pública da plataforma, permitindo seu uso por estudantes, pesquisadores
e educadores.



# **Referencial Teórico**

O projeto fundamenta-se na aplicação de técnicas de ciência de dados ao estudo
da literatura, integrando análise textual e análise de comportamento do usuário. A
utilização de sistemas de recomendação baseados em similaridade permite personalizar a
experiência do usuário, facilitando o acesso a conteúdos relevantes.

No contexto de séries temporais, os dados são analisados considerando sua
organização cronológica, permitindo identificar padrões como tendência, sazonalidade e
variações aleatórias. A autocorrelação mede a dependência entre observações ao longo
do tempo, enquanto funções como ACF e PACF auxiliam na identificação dessas
relações.

As médias móveis são utilizadas para suavizar a série temporal e evidenciar
padrões de longo prazo, enquanto a decomposição permite separar a série em
componentes interpretáveis, facilitando a análise dos fenômenos observados.

O projeto também se alinha ao ODS 4, ao promover o acesso à educação de
qualidade por meio de uma ferramenta digital voltada ao estudo da literatura. Sua
disponibilização como recurso acessível reforça seu caráter extensionista, contribuindo
para a democratização do conhecimento.

# **Pipeline da Solução**

A solução proposta será estruturada nas seguintes etapas:

a. Coleta de dados: Registro das interações dos usuários com marcação temporal.

b. Organização temporal: Estruturação dos dados como séries temporais.

c. Análise exploratória: Identificação de padrões iniciais e variações.

d. Autocorrelação: Análise das dependências entre observações ao longo do tempo.

e. Suavização: Aplicação de médias móveis para redução de ruído.

f. Decomposição: Separação da série em tendência, sazonalidade e resíduo.

g. Interpretação: Geração de insights para melhoria da plataforma.
