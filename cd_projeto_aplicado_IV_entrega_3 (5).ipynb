# **Introdução**

A literatura portuguesa constitui um importante patrimônio cultural, sendo
fundamental para a formação linguística, crítica e interpretativa de estudantes e
pesquisadores. No entanto, o acesso estruturado a obras literárias e a análise contextual
de seus conteúdos ainda representam desafios em ambientes digitais.

Nesse contexto, o projeto Memoria Litteraria consiste em uma aplicação web
voltada à busca textual e análise de obras da literatura portuguesa. A plataforma permite
localizar trechos específicos, analisar o uso de palavras em diferentes contextos e
recomendar obras e autores com base no perfil de uso. Seu público-alvo inclui estudantes,
pesquisadores, professores e escritores.

Além das funcionalidades de busca e recomendação, o sistema coleta dados  de
interação dos usuários, como consultas realizadas, funcionalidades utilizadas e
avaliações. Esses dados, organizados ao longo do tempo, configuram um cenário
adequado para análise por meio de séries temporais.

Séries temporais consistem em observações ordenadas cronologicamente,
permitindo a identificação de padrões como tendência, sazonalidade e variações
aleatórias. A análise desses padrões possibilita compreender a dinâmica de uso da
plataforma, identificar comportamentos recorrentes e apoiar a tomada de decisão baseada
em dados.

Dessa forma, o problema abordado neste projeto consiste na necessidade de
compreender como o comportamento dos usuários se desenvolve ao longo do tempo,
permitindo aprimorar a experiência na plataforma e orientar melhorias fundamentadas em
evidências.  

A motivação do projeto está na integração entre tecnologia e literatura, com o
objetivo de tornar o acesso ao conhecimento literário mais eficiente, estruturado e
orientado por dados. Do ponto de vista acadêmico, a análise de séries temporais permite
compreender padrões dinâmicos de comportamento, ampliando o uso de técnicas de
Ciência de Dados no contexto educacional. Sob a perspectiva tecnológica, a análise
temporal das interações dos usuários possibilita identificar padrões de uso, períodos de
maior atividade e mudanças de interesse, contribuindo para a evolução contínua da
plataforma.

No âmbito social, o projeto alinha-se ao Objetivo de Desenvolvimento Sustentável
4 (ODS 4), ao promover o acesso à educação de qualidade por meio de uma ferramenta
digital aberta e acessível. O caráter extensionista do projeto se manifesta na

# **Modelo base**

Após a etapa de análise exploratória e pré-processamento dos dados, foram aplicados diferentes modelos de previsão com o objetivo de estimar o consumo energético futuro e comparar seus desempenhos. A utilização de abordagens distintas permitiu avaliar tanto métodos estatísticos tradicionais quanto técnicas modernas baseadas em aprendizado profundo.

### ARIMA (AutoRegressive Integrated Moving Average)

O modelo ARIMA é uma das técnicas estatísticas mais utilizadas para previsão de séries temporais. Sua abordagem combina componentes autorregressivos (AR), diferenciação (I) e médias móveis (MA), permitindo modelar tendências e dependências temporais presentes nos dados.

A escolha desse modelo ocorreu devido à sua ampla utilização em problemas de previsão de demanda e consumo energético. O ARIMA é capaz de capturar padrões históricos da série temporal e gerar previsões com base no comportamento observado ao longo do tempo.

### Prophet

O Prophet é um modelo desenvolvido pelo Facebook (Meta) para previsão de séries temporais que apresentam tendências, sazonalidades e eventos recorrentes. Sua principal vantagem é a facilidade de modelagem de padrões temporais complexos sem a necessidade de extensos ajustes manuais.

No contexto deste projeto, o Prophet foi utilizado para analisar a capacidade de capturar comportamentos sazonais do consumo energético, oferecendo uma alternativa moderna aos modelos estatísticos tradicionais.

### LSTM (Long Short-Term Memory)

A arquitetura LSTM é uma variação das Redes Neurais Recorrentes (RNN) desenvolvida para lidar com dependências de longo prazo em séries temporais. Diferentemente dos modelos estatísticos tradicionais, a LSTM possui mecanismos internos de memória que permitem aprender padrões complexos presentes nos dados.

Esse modelo foi selecionado devido à sua elevada capacidade de identificar relações não lineares e comportamentos temporais mais sofisticados, frequentemente encontrados em séries de consumo energético. Além disso, a LSTM tem sido amplamente utilizada em aplicações de previsão de carga elétrica e demanda energética.

### Avaliação dos Modelos

Após o treinamento, os modelos foram avaliados por meio de métricas de desempenho amplamente utilizadas em problemas de regressão e previsão temporal. Entre elas destacam-se o Erro Médio Absoluto (MAE), a Raiz do Erro Quadrático Médio (RMSE) e o Erro Percentual Absoluto Médio (MAPE).

A comparação entre os resultados permitiu identificar qual abordagem apresentou melhor capacidade preditiva para os dados analisados, considerando tanto a precisão das previsões quanto a capacidade de generalização para dados não utilizados durante o treinamento.

### Comparação das Abordagens

Cada modelo apresenta características distintas. O ARIMA oferece uma abordagem estatística consolidada e de fácil interpretação, sendo adequado para séries temporais com padrões relativamente estáveis. O Prophet facilita a modelagem de tendências e sazonalidades, reduzindo a necessidade de ajustes complexos. Já a LSTM apresenta maior flexibilidade para aprender relações não lineares e padrões complexos, embora exija maior capacidade computacional e volume de dados para treinamento.

Dessa forma, a utilização conjunta dessas abordagens permitiu uma análise mais abrangente do problema, possibilitando a comparação entre técnicas clássicas e modernas de previsão de séries temporais aplicadas ao consumo energético.

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)

import numpy as np

def avaliar_modelo(nome, y_real, y_pred):

    mae = mean_absolute_error(
        y_real,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_real,
            y_pred
        )
    )

    mape = mean_absolute_percentage_error(
        y_real,
        y_pred
    )

    print(f"\n{nome}")
    print(f"MAE: {mae:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.4f}")

avaliar_modelo(
    "ARIMA",
    y_test,
    previsoes_arima
)

avaliar_modelo(
    "LSTM",
    y_test,
    previsoes_lstm
)
```
